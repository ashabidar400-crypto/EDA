```python
Q1=df['FM'].quantile(0.25)
Q3=df['FM'].quantile(0.75)
```

```python
IQR=Q3-Q1
ue=Q3+1.5*(IQR)
le=Q1-1.5*(IQR)
```

```python
df[df['FM']>ue]
```

```python
df[df['FM']>ue]=ue
```

```python
df.loc[df["FM"]>ue,"FM"]=ue
```

```python
df.loc[df["FM"]>le,"FM"]=le
```

```python
Q1=df['UC'].quantile(0.25)
Q3=df['UC'].quantile(0.75)
IQR=Q3-Q1
ue=Q3+1.5*(IQR)
le=Q1-1.5*(IQR)
df[df['UC']>ue]
df[df['UC']>ue]=ue
df.loc[df["UC"]>ue,"UC"]=ue
df.loc[df["UC"]>le,"UC"]=le
```
